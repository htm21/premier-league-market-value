# Data exploration

## Import libraries

```python
import pandas as pd
from pathlib import Path
import matplotlib.pyplot as plt
import seaborn as sns
```

## Load the data

```python
ROOT = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / "data").is_dir())
raw_data_dir = ROOT / "data" / "raw"
```

```python
competitions = pd.read_csv(raw_data_dir / "competitions.csv.gz")
clubs = pd.read_csv(raw_data_dir / "clubs.csv.gz")
valuations = pd.read_csv(raw_data_dir / "player_valuations.csv.gz")
players = pd.read_csv(raw_data_dir / "players.csv.gz")
appearances = pd.read_csv(raw_data_dir / "appearances.csv.gz")
```

### Get the Premier League appearances

```python
appearances[appearances["competition_id"] == PL_ID].head()
```

We want to focus on the 25/26 season

```python
appearances["date"] = pd.to_datetime(appearances["date"])
```

```python
appearances[(appearances["competition_id"] == PL_ID)]["date"].max()
```

Last game of the season was on 24 May 2026, so we will filter the appearances to only include those that happened between 1 July 2025 and 24 May 2026.

```python
pl_appearances = appearances[
    (appearances["competition_id"] == PL_ID)
    & (appearances["date"] >= pd.Timestamp("2025-07-01"))
    & (appearances["date"] <= pd.Timestamp("2026-05-24"))
].copy()

pl_appearances.tail()
```
